1.Generate Random Numbers and Find Even Numbers
Use NumPy's random module to generate 20 random integers between 1 and 100.

Write a program to:

Display the generated array. Find all the even numbers. Count how many even numbers were generated.

Concepts: np.random.randint(), filtering, len() / np.sum().


2.Generate Random Dice Rolls
Simulate rolling a six-sided dice 20 times using NumPy's random module.

Write a program to:

Generate 20 random dice results. Display the results. Count how many times the number 6 appeared. Display the percentage of rolls that resulted in 6.

Concepts: np.random.randint(), array filtering, counting, basic calculations.

```python
import numpy as np

arr = np.random.randint(1, 7, 20)

print("Generated array", arr)

count = arr[arr == 6]

length= len(count)

print("number of times 6 occurs", length)

percentage = length / len(arr) * 100

print("Percentage", percentage)
```

4. Random Number Guessing Game

Generate a random integer between 1 and 100 using NumPy's random module.

Ask the user to repeatedly enter a number until they guess the generated number.

After every attempt, display:

"Too High" if the user's guess is greater.
"Too Low" if the user's guess is smaller.
"Correct" when the user guesses the number.

Also display the number of attempts taken.

Use a while loop.

Concepts: np.random.randint(), while loop, user input, conditional statements.

```python
x=np.random.randint(1,101)

attempt=0

while True:
   guess=int(input("Enter a number"))
   attempt=attempt+1

   if (guess>x):
     print("Too High:(,Try Again")
   elif(guess<x):
     print("Too Low :(,Try Again)")
   else:
      print("Congratulations!!!,It's the correct answer")
```

5. Random Sales Data Generator and Analysis

A company wants to generate sample sales data for 30 days.

Write a program to:

Generate the sales data.
Calculate the total sales.
Calculate the average daily sales.
Find the highest and lowest sales.
Identify all days where sales were above the average.

Using a for loop, display each day and its sales in the format:
Day 1 : ₹xxxxx
Day 2 : ₹xxxxx
...

Concepts: random data generation, arrays, for loop, indexing, filtering, aggregation, argmax().


```python
sales=np.random.randint(5000,50001,30)
print("Sales Data",sales)

Total=np.sum(sales)
print("Total Sales",Total)

Average=np.mean(sales)
print("Average Sales",Average)

high=np.max(sales)
print("Highest Sales",high)

low=np.min(sales)
print("Lowest Sales",low)
above=sales[sales>Average]
print("Sales above Average",len(above))

print("the day/index with the highest sales:",np.argmax(sales)+1)

count=sales[sales>30000]
print("Count how many days had sales above ₹30,000:",count)


for i in range(len(sales)):
   print("Day",i+1,": ₹",sales[i])
```

6. Random Password/OTP Security Simulation

Generate 100 random 6-digit OTPs using NumPy's random module.

Write a program to:

Generate 100 six-digit OTPs.
Check how many unique OTPs were generated.
Identify whether any OTP was generated more than once.
Find the repeated OTPs, if any.
Count the number of OTPs that contain at least one repeated digit.

For example:

123456 → No repeated digit
112345 → Repeated digit
778899 → Repeated digits

Use a for loop to examine the digits of each OTP.

Concepts: np.random.randint(), loops, uniqueness, indexing, string/list manipulation, np.unique(), conditional logic.

```python
import numpy as np
otp=np.random.randint(100000,1000000,100)
print(otp)

unique=np.unique(otp)
unique_count=len(unique)
print("Uniques otp generated",unique_count)

repeated_otp=[]

for i in unique:
   count=np.sum(otp==i)

   if count>1:
     repeated_otp.append(i)

if len(repeated_otp)>0:
     print("Repeated Values are present")
else:
    print("No Duplicates")

repeated_digit_count=0
for i in otp:
  digit=str(i)

  if len(set(digit))<len(digit):
      repeated_digit_count+=1

print("Repeated digit", repeated_digit_count)
```
